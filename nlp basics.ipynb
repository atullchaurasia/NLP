lemmatization

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
import string
lemmatized_sentences = []

for text in data['statement']:
    words = word_tokenize(text)
    words = [word for word in words if word not in string.punctuation]
    stemmed_words = [lemmatizer.lemmatize(word) for word in words]
    stemmed_sentence = ' '.join(stemmed_words)
    lemmatized_sentences.append(stemmed_sentence)
    
data['lemmatized'] = lemmatized_sentences
```

stemming

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
```

```python
import string

stemmed_sentences = []

for text in data['statement']:
    words = word_tokenize(text)
    words = [word for word in words if word not in string.punctuation]
    stemmed_words = [stemmer.stem(word) for word in words]
    stemmed_sentence = ' '.join(stemmed_words)
    stemmed_sentences.append(stemmed_sentence)
data['stemming'] = stemmed_sentences  
```

N-gram extraction(unigram, bigram, tigram)

```python
from nltk.util import ngrams


unigram_list = []
bigram_list = []
trigram_list = []


for text in(data['statement']):
    words = word_tokenize(text)
    
    unigram = list(ngrams(words, 1))
    bigram = list(ngrams(words, 2))
    trigram = list(ngrams(words, 3))
    
    unigram_list.append(' '.join([' '.join(gram) for gram in unigram]))
    bigram_list.append(' '.join([' '.join(gram) for gram in bigram]))
    trigram_list.append(' '.join([' '.join(gram) for gram in trigram]))
```

```python
data['unigram'] = unigram_list
data['bigram'] = bigram_list
data['trigram'] = trigram_list
```

```python
from sklearn.feature_extraction.text import CountVectorizer

def generate_ngrams(text_column, n):
    vectorizer = CountVectorizer(ngram_range=(n, n), stop_words='english')
    ngram_matrix = vectorizer.fit_transform(text_column)
    ngram_features = vectorizer.get_feature_names_out()
    ngram_counts = ngram_matrix.sum(axis=0).tolist()[0]

    ngram_df = pd.DataFrame({"ngram": ngram_features, "count": ngram_counts})
    ngram_df = ngram_df.sort_values(by="count", ascending=False)
    return ngram_df


for n in [1, 2, 3]:
    print(f"\nTop {n}-grams:")
    ngram_df = generate_ngrams(data["statement"], n)
    print(ngram_df.head(10))
```

Name Entity Recognition NER

```python
import spacy
nlp = spacy.load("en_core_web_sm")

named_entities_list = []
for statement in data['statement']:
  statement = str(statement)
  doc = nlp(statement)
  entities = [ent.text for ent in doc.ents]
  named_entities_list.append(", ".join(entities) if entities else "No entities")

data['Named entities'] = named_entities_list
```
